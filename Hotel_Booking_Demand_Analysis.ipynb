### 4.2.1 Handling Missing Values in Children 
The `children` column contains 4 missing values. Since the number of missing observations is very small compared with the size of the dataset, removing these records would result in unnecessary data loss.

The missing values will therefore be replaced using the median of the available `children` values. The median is used because it is less sensitive to extreme values and is appropriate for this count-based variable.

```python
df["children"]=df["children"].fillna(df["children"].median())
print("Missing children values:", df["children"].isnull().sum())
```

### Findings
The **4** missing values in the `children` column were replaced using the median value of available observations. The verification check returned 0 missing values, confirming that the missing entries were successfully handled without removing any records from the dataset.

### 4.2.2 Handling Missing Values in Country
The `country` column contains **452** missing values. Since the column represents the country of origin of the guest, replacing the missing values with a specific country could introduce inaccurate information into the dataset.

The missing values will therefore be represented as `Unknown` rather than being removed or assigned an assumed country. This preserves the booking records while clearly distinguishing missing information from known countries.

```python
df["country"].isnull().sum()
```

```python
df["country"] =df["country"].fillna("Unknown")
```

```python
df["country"].isnull().sum()
```

### Findings
The 452 missing values in the `country` column were replaced with `Unknown`. This approach preserved the affected booking records without assigning an unsupported country to guests whose country information was unavailable.
